```python
# reverse a string 
str=input("Enter bString :: ")
rev=""

for ch in str:
    rev=ch+rev

print("Reveresed String is :: ",rev)
```

```python
# find occurance of character 
str=input("Enter bString :: ")
max_char=""
max_count=0

for ch in str:
    count=0
    for c in str:
        if ch==c:
            count=count+1

    if count>max_count:
        max_count=count
        max_char=ch

print("Maximum Ocuuring Character is :: ",max_char)
print("Maximum  Character count is :: ",max_count)
```

```python
# count the occurance of given character
str=input("Enter String :: ")
ch=input("Enter Character to find Ocuurance :: ")
count=0

for c in str.lower():
    if c==ch:
        count+=1

print("Occurance of given character is :: ",count)
```

```python
# remove the spaces from string
str=input("Enter String :: ")
result=""

for ch in str:
    if ch!=" ":
        result+=ch

print("String without Spaces is :: ",result)
```

```python
# copy one string to other string
str=input("Enter String :: ")
copy_str=""

for ch in str:
    copy_str+=ch

print("Original string is :: ",str)
print("Copied String is :: ",copy_str)
```

```python
# remove the duplicates 

str=input("Enter String :: ")
res_str=""

for ch in str:
    if ch not in res_str:
        res_str+=ch

print("String without Duplication is :: ",res_str)
```

```python
# compare two str

s1=input("Enter String :: ")
s2=input("Enter String :: ")

if len(s1)!=len(s2):
    print("String are not equal ")
else:
    equal=True
    for i in range(len(s1)):
        if s1[i]!=s2[i]:
            equal=False
            break

    if equal:
        print("String is equal")
    else:
        print("String is Not Equal ")
```
