```python
# join the strings
words=["India","is","my","country"]

res=" ".join(words)
print("String is :: ",res)

res="-".join(words)
print("String is :: ",res)

res=":".join(words)
print("String is :: ",res)

res=",".join(words)
print("String is :: ",res)

res="|".join(words)
print("String is :: ",res)

res="=".join(words)
print("String is :: ",res)
```

```python
# check whether my string contains only digit

str=input("enter string :: ")

if str.isdigit():
    print("String contains Digit")
else:
    print("It is a Pure String ")
```

```python
# check whether my string contains only characters

str=input("enter string :: ")

if str.isalpha():
    print("String contains Alphabets")
else:
    print("It contains numbers also  ")
```

```python
# check whether my string contains digits and  characters

str=input("enter string :: ")

if str.isalnum():
    print("String contains Alphabets and digits")
else:
    print("It contains numbers or characters only  ")
```

```python
# check string starts or ends with some specific word 

str=input("enter string :: ")
sword=input("Enter Word to check the starting word :: ")

if str.startswith(sword):
    print("Yes String Starts with the given word ")
else:
    print("No String Don't Starts with the given word ")


eword=input("Enter Word to check the ending word :: ")

if str.endswith(sword):
    print("Yes String ends with the given word ")
else:
    print("No String Don't ends with the given word ")
```

```python
# calcualte the count of vowels 

str=input("enter string :: ")
count=0

for ch in str.lower():
    if ch in "aeiou":
        count=count+1

print("Total Vowel COunt is :: ",count)
```
