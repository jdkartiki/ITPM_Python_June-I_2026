```python
# create a grade system

marks=float(input("Enter Marks :: "))

if marks>=85 and marks<=100:
    print("Grade A")
elif marks>=60 and marks<85:
    print("Grade B")
elif marks>=35 and marks<60:
    print("Grade C")
else:
    print("Fail")
```

```python
# accept the colour of signal and give output
signal=input("Enter Colour of Signal :: ")

if signal=="red":
    print("Stop")
elif signal=="green":
    print("Go")
elif signal=="yellow":
    print("Slow Down The Vehicle")
else:
    print("Invalid Signal Colour ")
```

```python
# check the customer membership -- premium,gold,regular

bal=int(input("Enter Balance :: "))

if bal>=1000000:
    print("Premium Membership")
elif bal<1000000 and bal>=500000:
    print("Gold Membership")
else:
    print("Regular Membership")
```

```python
# accept the bill and based on that give discount

bill=int(input("Enter bill amount :: "))

if bill>=10000:
    discount=bill*0.25
    bill=bill-discount
    print("Discount is :: ",discount)
    print("Bill Amount :: ",bill)
elif bill<10000 and bill>=5000:
    discount=bill*0.15
    bill=bill-discount
    print("Discount is :: ",discount)
    print("Bill Amount :: ",bill)
elif bill<5000 and bill>=2000:
    discount=bill*0.5
    bill=bill-discount
    print("Discount is :: ",discount)
    print("Bill Amount :: ",bill)
else:
    print("No Discount!!! ")
    print("Bill Amount :: ",bill)
```

```python
# accept the salary and find the tax slab for it

sal=float(input("Enter Salary :: "))

if sal<=500000:
    print("No TAX")
elif sal<=800000 and sal>500000:
    print("10 % Tax ")
elif sal<=1200000 and sal>800000:
    print("15% Tax")
else:
    print("20% Tax")
```
