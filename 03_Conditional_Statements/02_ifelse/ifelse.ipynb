```python
# num -- positive or negative

num=int(input("Enter number :: "))

if num>=0:
    print("Positive Number ")
else:
    print("Negative Number")
```

* = is used for assignment
    * 1. eg : a=20,b=40 

* == is used for comparision of values only 
    * 1. a=20 b=20 -- true
    * 2. a=20 b='20' -- true

* === is used for comparision of value and datatype both 
    * 1. a=20 b=20 -- true
    * 2. a=20 b-'20' -- false 

```python
# username =admin password=admin@1234

username=input("Enter Username :: ")
password=input("Enter Password :: ")

if username=='admin' and password=='admin@1234':
    print("Login Successfulll ")
else:
    print("Login Failed ")
```

```python
# check the strength of password 

password=input("Enter Password :: ")

if len(password)>=8:
    print("Strong Password ")
else:
    print("Weak Password ")
```

```python
# check weekend sat or sun 
# var are case sensitive 

day=input("Enter Day name :: ")

if day=='sat' or day=='sun':
    print("Its WEEEKEND !!!")
else:
    print("Its a WEEKDAY ")
```
