```python
# create login system 

username=input("Enter Username :: ")
password=input("Enter Password :: ")

if username=="admin":
    if password=="admin@1234":
        print("Login Successfull ")
    else:
        print("Invalid Password")
else:
    print("Invalid Username ! Login Failed ")
```

```python
# accept bill amount and give the shopping offer

bill=float(input("Enter Bill AMount :: "))
prime_mem=input("Enter Prime Member :: yes/no")

if bill>10000:
    if prime_mem=="yes":
        print("Extra 10% off ")
    else:
        print("5 % Off ")
else:
    print("No Discount !!! ")
```

```python
# accept salary and exp from emp and check bonus eligibility 

sal=float(input("Enter salary :: "))
exp=int(input("Enter Experience in Years:: "))

if sal>=50000:
    if exp>5:
        print("Eligible for Bonus ")
    else:
        print("Less Than 5 Yesar of Experienced Candidates are not eligible for bonus")
else:
    print("Salary is less than required sal for bonus")
```
