```python
# print sqaure pattern of 5*5

for i in range(5):
    for j in range(5):
        print("*",end=" ")
    print()
```

```python
# right angles triangle using * 

for i in range(1,6):
    for j in range(i):
        print("*",end=" ")
    print()
```

```python
# right angle triangle but using numbers 

for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# right angle triangle but using numbers 

for i in range(1,6):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
```

```python
# reverse pattern 

for i in range(5,0,-1):
    for j in range(i):
        print("*",end=" ")
    print()
```

```python
# reverse number pattern 
for i in range(5,0,-1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# reverse number pattern 
for i in range(5,0,-1):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
```

```python
# same rows and columns number triangle
for i in range(1,6):
    for j in range(i):
        print(i,end=" ")
    print()
```

```python
# alpahabet triangle pattern

for i in range(1,6):
    for j in range(i):
        print(chr(64+i),end=" ")
    print()
```

```python
# print continous number pattern 

num=1

for i in range(1,6):
    for j in range(i):
        print(num,end=" ")
        num=num+1
    print()
```

```python
# print the marksheet of student
students=["Rahul","Raj","Ram"]

for i in students:
    print("Student NAme is :: ",i)

    for subject in ["Java","Python","SQL"]:
        print("Subject is :: ",subject)

    print()
```

```python
# cinema seating arrangements

for row in range(1,4):
    print("Row number :: ",row)

    for seats in range(1,6):
        print("Seat No. ::",seats,end="\t")

    print("\n")
```
