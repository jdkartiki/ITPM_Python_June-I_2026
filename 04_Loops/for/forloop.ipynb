range():
* range() is used to give a particular range of number 
* is inbuilt function
* range(start number,ending number+1)
* range function will take ending number only upto that number 
* eg: range(1,10) --- 9 times  range(1,11)-- 10 times 

```python
# print 1 to 10 
# for( i=1;i<=10;i++)

for i in range(1,11):
    print(i)
```

```python
# accept number from user and calculate table of it 

num=int(input("Enter number :: "))

for i in range(1,11):
    print(num," * ",i," = ",num*i)
```

```python
# factorial 
num=int(input("Enter Number to calculate factorial :: "))
fact=1

for i in range(1,num+1):
    fact=fact*i

print("Factorial is :: ",fact)
```

```python
# sum of first 5 numbers 

num=5
sum=0

for i in range(1,num+1):
    sum=sum+i

print("Sum is :: ",sum)
```

```python
# calculate the total salaries of emp in company

salaries=[20000,40000,50000,20000,10000]
total_sal=0

for i in salaries:
    total_sal=total_sal+i

print("Total Salary is :: ", total_sal)
```
