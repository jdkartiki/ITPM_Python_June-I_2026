```python
# sum of numbers

numrange=int(input("Enter range of numbers to add :: "))

i=1
sum=0

while i<=numrange:
    sum=sum+i
    i=i+1

print("Sum is :: ",sum)
```

```python
# ATM Correct Pin Verification -- 3 times 

correct_pin=2026
attempt=1

while attempt<=3:
    pin=int(input("Enter Pin :: "))

    if pin==correct_pin:
        print("Login Successfull !!! ")
        break
    print("Wrong Pin!!!")
    attempt+=1

if attempt>3:
    print("Account Blocked !!!!")
```

```python
# shopping cart 

cart=[]
item=""

while item!="done":
    item=input("Enter Item Name (or done to finish shopping ) :: ")

    # append() is used to add elements in list/array
    if item!="done":
        cart.append(item)

print("Items in Cart is :: ",cart)
```

```python
# attendence system 

name=""
count=0

while name!="stop":
    name=input("Enter Student Name :: ")

    if name!="stop":
        print(name," is present ")
        count=count+1

print("Attendence is Completed ")
print("Count is :: ",count)
```
