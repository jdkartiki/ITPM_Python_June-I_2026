```python
# atm machine

choice=int(input("Enter Your choice :: \n1.Balance Enquiry\n2.Deposit \n3.Withdraw"))

match choice:
    case 1:
        print("Balance Check")
    case 2:
        print("Deposit Money")
    case 3:
        print("Withdraw Money")
    case _:
        print("Invalid Choice")
```

```python
# check the traffic signals

signal=input("Enter colour of signal :: ")

match signal:
    case "red":
        print("STOP")
    case "green":
        print("GO")
    case "yellow":
        print("WAIT")
    case _:
        print("Invalid Signal ")
```

```python
# emp role 

emp_role=input("Enter ROle :: ")

match emp_role:
    case "manager":
        print("Employee is manager ")
    case "developer":
        print("Employee is Developer")
    case "Tester":
        print("EMployee us Tester")
    case "HR":
        print("Employee is HR")
    case _:
        print("Invalid Input ")
```

```python
# food ordering system 

itemname=input("Enter Food Name :: [pizza/burger/sandwitch] ")

match itemname:
    case "pizza":
        print("Price is :: Rs. 300")
    case "burger":
        print("Price is :: Rs. 150")
    case "sandwitch":
        print("Price is :: Rs. 100")
    case _:
        print("Please Select itemname ")
```
