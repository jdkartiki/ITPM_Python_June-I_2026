```python
# skip number 5

for i in range(1,11):
    if i==5:
        continue
    print(i)
```

```python
# skip even number
for i in range(1,11):
    if i%2==0:
        continue
    print(i)
```

```python
# skip negative numbers
nums=[2,0,-2,-4,-6,0,9]

for i in nums:
    if i<0:
        continue
    print(i)
```

```python
# skip the faield stduents

marks=[10,20,40,45,60,9,0]

for i in marks:
    if i<35:
        continue
    print("Pass Student AMrks :: ",i)
```
