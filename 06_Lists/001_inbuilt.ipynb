 1. What is a List?

A **list** in Python is a collection of ordered elements enclosed in square brackets `[ ]`.

* Lists can store multiple data types
* Lists are **mutable** (can be changed)

 Example

```python
numbers = [10, 20, 30, 40]
mixed = [1, "Python", 3.5, True]
```

 4. List Indexing

 Positive Indexing

```python
lst = [10, 20, 30, 40]
print(lst[0])   # 10
print(lst[2])   # 30
```

 Negative Indexing

```python
print(lst[-1])  # 40
print(lst[-3])  # 20
```

Indexing Diagram

```
Index:   0    1    2    3
Value:  10   20   30   40
Neg:    -4   -3   -2   -1
```

---

 5. List Slicing

 Syntax

```python
list[start : end : step]
```

 Examples

```python
lst = [10, 20, 30, 40, 50]

print(lst[1:4])   # [20, 30, 40]
print(lst[:3])    # [10, 20, 30]
print(lst[::2])   # [10, 30, 50]
print(lst[::-1])  # Reverse list
```

---

 6. Modifying List Elements (Mutability)

 Change Element

```python
lst = [10, 20, 30]
lst[1] = 99
print(lst)
```

 Add Elements

```python
lst.append(40)
lst.insert(1, 15)
```

 Remove Elements

```python
lst.remove(20)
lst.pop()
```

### 7. Common List Methods

| Method    | Description           |
| --------- | --------------------- |
| append()  | Add element at end    |
| insert()  | Add at specific index |
| extend()  | Add multiple elements |
| remove()  | Remove specific value |
| pop()     | Remove by index       |
| clear()   | Remove all elements   |
| index()   | Find position         |
| count()   | Count occurrences     |
| sort()    | Sort list             |
| reverse() | Reverse list          |

 Example

```python
lst = [3, 1, 4, 2]
lst.sort()
print(lst)
```

 8. Traversing Lists

 *  Using for Loop

```python
lst = [10, 20, 30]
for i in lst:
    print(i)
```

 * Using while Loop

```python
lst = [10, 20, 30]
i = 0

while i < len(lst):
    print(lst[i])
    i += 1
```

 9. List Operators

 * Concatenation

```python
a = [1, 2]
b = [3, 4]
print(a + b)
```

 * Repetition

```python
print([1, 2] * 3)
```

 * Membership

```python
print(20 in [10, 20, 30])
```

 10. List Comprehension

 Theory

List comprehension provides a short and efficient way to create lists.

 Syntax

```python
[expression for item in iterable if condition]
```

 Examples

```python
squares = [x*x for x in range(5)]
even = [x for x in range(10) if x % 2 == 0]
```

 11. Nested Lists

 Example

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

for row in matrix:
    for val in row:
        print(val, end=" ")
```

 14. Important List Programs (Exam)

 Sum of List Elements

```python
lst = [10, 20, 30]
print(sum(lst))
```

```python
lst = [5, 8, 2]
print(max(lst))
```
